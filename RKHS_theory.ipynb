### RKHS

##### Riesz representation theorem

(Rudin 1991, theorem 12.5)

There is a linear isometry $y \mapsto \Lambda$ of $\mathbb{H}$ onto $\mathbb{H}^*$ given by $\Lambda x = \langle x, y \rangle$. 

##### point-evaluation functional

for a Hilbert space $\mathbb{H}$ of real-valued functions defined on the set $E$, for each $z \in E$ we can consider a point-evaluation functional $L_z(f) = f(z)$.

##### Definition 3.1 (Yun):

for a hilbert space $\mathbb{H}$ of real-valued functions defined on the set $E$, 

we say that $K: E \times E \to \mathbb{R}$ is a reproducing kernel for $\mathbb{H}$ if the following hold:
- At each point $z \in E$, some point-evaluation generator $k_z(\cdot) = K(\cdot, z)$ exists in $\mathbb{H}$.
- K satisfies the reproducing property: for any $f \in \mathbb{H}$, point evaluation at $z \in E$ is given by $f(z) = \langle f, k_z \rangle _\mathbb{H}$.

A hilbert space equipped with a reproducing kernel is a RKHS.

### Moore-Aronsajz theorem

Reproducing kernels are symmetric and positive semi-definite.

Conversely, any symmetric positive (semi?)-definite kernel induces a unique RKHS, denoted $\mathbb{H}(K)$.

### Representer Theorem

For an empirical risk functional $f \mapsto E(S) + g(||f||)$ defined on a RKHS $H_k$, consisting of:
- a training sample $ S = (x_1, y_1), \cdots, (x_n, y_n)$,
- an error function $E((x_1, y_1, f(x_1)), (x_n, y_n, f(x_n)))$,
- a strictly-increasing regularization function $g: \left[0, \infty\right) \to \mathbb{R}^n$,

the minimizer $f^*$ can be represented as $f^*(\cdot) = \sum_{i=1}^n \alpha_i k(\cdot, x_i)$ for some $\alpha \in \mathbb{R}^n$.

In effect, (depending on the risk functional,) this reduces finding the minimizer to a linear problem.

As you can probably guess, in our X-ray reconstruction setup, the training sample will be the approximate sinogram constructed from limited projection angles.
